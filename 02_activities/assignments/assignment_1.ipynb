### Part 1: Building the base Anagram Checker

Given two valid strings, check to see if they are anagrams of each other. If it is, return `True`, else `False`. For this part, we can assume that uppercase letters are the same as if it was a lowercase character.

Examples of anagrams:
* Silent and Listen
* Night and Thing

Example outputs:
```python
anagram_checker("Silent", "listen") # True
anagram_checker("Silent", "Night") # False
anagram_checker("night", "Thing") # True
```

````python
# For testing purposes, we will write our code in the function

from itertools import permutations # import package to use permutations function

'''
Function name:  anagram_checker
Objective:      Checks if the two input words are anagrams of each other, case does not matter
Inputs:         word_a, word_b
Output:         Returns True for anagrams detected; returns False for no anagrams detected
'''

def anagram_checker(word_a, word_b):
  # Your code here
  list_a = list(word_a.lower()) # Convert word_a into lower case and turn it into a list
  word_b = word_b.lower() # Convert word_b into lower case and replace it

  anagram_val = 0 # Set default value of anagram_val as 0 to signify no anagrams are detected

  # For loop to run through all permutations of lower case of word_a and compare each iteration to word_b
  # If a permutation of word_a matches word_b, variable called anagram_val is set to 1 to signify anagrams are detected
  for p in permutations(list_a):
    string_a = ''.join(p)
    
    if word_b == string_a:
      anagram_val += 1
      exit

  # Print boolean value of anagram_val
  print(bool(anagram_val))

# Run your code to check using the words below:
anagram_checker("Silent", "listen")
````

````python
anagram_checker("Silent", "Night")
````

````python
anagram_checker("night", "Thing")
````

### Part 2: Expanding the functionality of the Anagram Checker

Using your existing and functional anagram checker, let's add a boolean option called `is_case_sensitive`, which will return `True` or `False` based on if the two compared words are anagrams and if we are checking for case sensitivity.

````python
# import package to use permutations function
# Repeating here in case the cell containing the function above was not run
from itertools import permutations

'''
Function name:  anagram_checker
Objective:      Checks if the two input words are anagrams of each other, case matters
Inputs:         word_a, word_b, is_case_sensitive
Output:         Returns True for anagrams detected; returns False for no anagrams detected
'''

def anagram_checker(word_a, word_b, is_case_sensitive):
  # Modify your existing code here
  
  # If input is_case_sensitive is True, compare words as-is
  # If input is_case_sensitive is False, compare words in their lowercase
  if is_case_sensitive == False:
    list_a = list(word_a.lower()) # Convert word_a into lower case and turn it into a list
    word_b = word_b.lower() # Convert word_b into lower case
  else:
    list_a = list(word_a) # Turn word_a as-is into a list

  anagram_val = 0 # Set default value of anagram_val as 0 to signify no anagrams are detected

  # For loop to run through all permutations of word_a and compare each iteration to word_b
  # If a permutation of word_a matches word_b, variable called anagram_val is set to 1 to signify anagrams are detected
  for p in permutations(list_a):
    string_a = ''.join(p)
    
    if word_b == string_a:
      anagram_val += 1
      exit

  # Print boolean value of anagram_val
  print(bool(anagram_val))
  

# Run your code to check using the words below:
anagram_checker("Silent", "listen", False) # True
````

````python
anagram_checker("Silent", "listen", True) # False
````
